```python
cities = [
    "Hyderabad",
    "Mumbai",
    "Delhi",
    "Bengaluru",
    "Chennai",
    "Kolkata",
    "Pune",
    "Ahmedabad",
    "Jaipur",
    "Lucknow"
]

regions = {
    "Hyderabad": "South",
    "Mumbai": "West",
    "Delhi": "North",
    "Bengaluru": "South",
    "Chennai": "South",
    "Kolkata": "East",
    "Pune": "West",
    "Ahmedabad": "West",
    "Jaipur": "North",
    "Lucknow": "North"
}
```

```python
warehouse_data = []
```

```python
for i, city in enumerate(cities, start=1):

    warehouse_id = f"W{i:03}"

    warehouse_name = city + " Warehouse"

    region = regions[city]

    capacity = random.randint(3000, 10000)

    warehouse_data.append([
        warehouse_id,
        warehouse_name,
        city,
        region,
        capacity
    ])
```

```python
warehouses_df = pd.DataFrame(
    warehouse_data,
    columns=[
        "Warehouse_ID",
        "Warehouse_Name",
        "City",
        "Region",
        "Capacity"
    ]
)

warehouses_df
```

```python
warehouses_df.to_csv("../Dataset/Warehouses.csv", index=False)

print("Warehouses.csv created successfully!")
```

```python
product_data = []

product_id = 101

for category in catalog:

    brands = catalog[category]["brands"]
    products = catalog[category]["products"]

    for brand in brands:

        for product in products:

            unit_cost = random.randint(500, 50000)

            unit_price = int(unit_cost * random.uniform(1.15, 1.40))

            product_data.append([
                product_id,
                brand + " " + product,
                category,
                brand,
                unit_cost,
                unit_price
            ])

            product_id += 1
```

```python
products_df = pd.DataFrame(
    product_data,
    columns=[
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Unit_Cost",
        "Unit_Price"
    ]
)

products_df
```

```python
from datetime import datetime, timedelta
```

```python
start_date = datetime(2025, 1, 1)
```

```python
orders_data = []
```

```python
order_id = 1

for i in range(50000):

    product = products_df.sample(1).iloc[0]

    warehouse = random.choice(warehouse_ids)

    quantity = random.randint(1, 10)

    order_date = start_date + timedelta(days=random.randint(0, 365))

    unit_price = product["Unit_Price"]

    total_amount = quantity * unit_price

    orders_data.append([
        f"O{order_id:06}",
        order_date.strftime("%Y-%m-%d"),
        product["Product_ID"],
        warehouse,
        quantity,
        unit_price,
        total_amount
    ])

    order_id += 1
```

```python
orders_df = pd.DataFrame(
    orders_data,
    columns=[
        "Order_ID",
        "Order_Date",
        "Product_ID",
        "Warehouse_ID",
        "Quantity",
        "Unit_Price",
        "Total_Amount"
    ]
)

orders_df
```

```python
orders_df.to_csv("../Dataset/Orders.csv", index=False)

print("Orders.csv created successfully!")
```
